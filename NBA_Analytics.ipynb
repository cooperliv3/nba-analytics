```python
import pandas as pd
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/fivethirtyeight/data/master/nba-elo/nbaallelo.csv"

response = requests.get(url)
df = pd.read_csv(StringIO(response.text))

df.head()
```

```python
#Which team has the most amount of wins in the dataset?
df['winning_team'] = df['pts'] > df['opp_pts']
wins_per_team = df[df['winning_team']].groupby('team_id').size().sort_values(ascending = False)
wins_per_team.head()
```

```python
#Since 2000, which teams have been the most dominant?
df_filtered = df[(df['year_id'] >= 2000) & (df['year_id'] <= 2015)]

df_filtered['winning_team'] = df_filtered['pts'] > df_filtered['opp_pts']

wins = df_filtered[df_filtered['winning_team']]

fran_wins = wins.groupby('fran_id').size().reset_index(name='Wins Since 2000')

fran_wins = fran_wins.rename(columns={'fran_id': 'Team'})

fran_wins.head(30)
```

```python
#How many wins does the Thunder franchise have all time?
df['winning_team'] = df['pts'] > df['opp_pts']

team = 'OKC'

team_wins = df[(df['team_id'] == team) & (df['winning_team'])].shape[0]

team_summary = pd.DataFrame([{'Team': team, 'All Time Wins': team_wins}])

team_summary.head()
```

```python
#How many wins does the Thunder have each year?
# Determine if the team won the game
df['winning_team'] = df['pts'] > df['opp_pts']

# Filter for OKC wins
okc_wins_by_year = df[(df['team_id'] == 'OKC') & (df['winning_team'])]

# Group by year_id and count wins
wins_per_year = okc_wins_by_year.groupby('year_id').size().reset_index(name='Wins')

# Rename column for display
wins_per_year = wins_per_year.rename(columns={'year_id': 'Year'})

wins_per_year.head()
```
